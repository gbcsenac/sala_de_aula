```python
class Cliente:
    def __init__(self, nome: str, telefone: str, email: str, tipo: str, id_cliente: int = None):
        self.id_cliente = id_cliente
        self.nome = nome
        self.telefone = telefone
        self.email = email
        self.tipo = tipo
    
    def __str__(self):
        return f"Cliente(nome={self.nome}, telefone={self.telefone}, email={self.email}, tipo={self.tipo})"
    
    def __repr__(self):
        return (f"Cliente(id_cliente={self.id_cliente!r}, " f"nome={self.nome!r}," f"telefone={self.telefone!r}," f"email={self.email!r}," f"tipo={self.tipo!r})"

     )
```

```python
#Cadastrar cliente
def inserir_cliente(self, conexao):
    cursor = None

    try:
        cursor = conexao.cursor()

        sql = """INSERT INTO cliente (nome, telefone, email, tipo)
                 VALUES (%s, %s, %s, %s)
                 RETURNING id_cliente;"""

        valores = (self.nome, self.telefone, self.email, self.tipo)

        cursor.execute(sql, valores)

        self.id_cliente = cursor.fetchone()[0]

        conexao.commit()

        print("Cliente cadastrado com sucesso!")

        return self.id_cliente

    except Exception as erro:
        conexao.rollback()

        print(f"Erro ao cadastrar cliente: {erro}")

        return None

    finally:
        if cursor is not None:
            cursor.close()
```

```python
#Listar clientes
@staticmethod
def listar_clientes(conexao):
    cursor = None

    try:
        cursor = conexao.cursor()

        sql = """SELECT id_cliente, nome, telefone, email, tipo
                 FROM cliente
                 ORDER BY id_cliente;"""

        cursor.execute(sql)
        resultados = cursor.fetchall()

        clientes = []

        for registro in resultados:
            cliente = Cliente(
                nome=registro[1],
                telefone=registro[2],
                email=registro[3],
                tipo=registro[4],
                id_cliente=registro[0]
            )

            clientes.append(cliente)

        return clientes

    except Exception as erro:
        print(f"Erro ao listar clientes: {erro}")
        return []

    finally:
        if cursor is not None:
            cursor.close()
```

```python
#Atualizar cliente
def atualizar_cliente(self, conexao):
    cursor = None

    try:
        cursor = conexao.cursor()

        sql = """UPDATE cliente
                  SET nome = %s, telefone = %s, email = %s, tipo = %s
                  WHERE id_cliente = %s;"""

        valores = (self.nome, self.telefone, self.email, self.tipo, self.id_cliente)

        cursor.execute(sql, valores)

        if cursor.rowcount == 0:
            conexao.rollback()
            print("Cliente não encontrado")
            return False
        
            conexao.commit()
            print("Cliente atualizado com sucesso!")
            return True

    except Exception as erro:
        conexao.rollback()

        print(f"Erro ao atualizar cliente: {erro}")

        return False

    finally:
        if cursor is not None:
            cursor.close()
```

```python
#Excluir cliente
@staticmethod
def excluir_cliente(id_cliente: int, conexao):
    cursor = None

    try:
        cursor = conexao.cursor()

        sql = """DELETE FROM cliente WHERE id_cliente = %s"""

        cursor.execute(sql, (id_cliente,))

        if cursor.rowcount == 0:
            conexao.rollback()
            print("Cliente não encontrado")
            return False

        conexao.commit()
        print("Cliente excluído com sucesso!")
        return True

    except Exception as erro:
        conexao.rollback()
        print(f"Erro ao excluir cliente: {erro}")
        return False

    finally:
        if cursor is not None:
            cursor.close()
```

```python
class Produto:
    id_produto: int
    nome: str
    tipo: str
    preco: float
    quantidade_disponivel: int

    def __init__(self, nome: str, tipo: str, preco: float, quantidade_disponivel: int):
        self.id_produto = None
        self.nome = nome
        self.tipo = tipo
        self.preco = preco
        self.quantidade_disponivel = quantidade_disponivel

    def __str__(self):
        return f"Produto(id_produto={self.id_produto}, nome={self.nome}, tipo={self.tipo}, preco={self.preco}, quantidade_disponivel={self.quantidade_disponivel})"

    def __repr__(self):
        return self.__str__()
```

```python
#inserir produto
def inserir_produto(self,conexao):
    cursor = None
    try:
        cursor = conexao.cursor()

        sql = """INSERT INTO produto (nome, tipo, preco, quantidade_disponivel)
                 VALUES (%s, %s, %s, %s)
                 RETURNING id_produto;"""

        valores = (self.nome, self.tipo, self.preco, self.quantidade_disponivel)

        cursor.execute(sql, valores)

        self.id_produto = cursor.fetchone()[0]

        conexao.commit()

        print("Produto cadastrado com sucesso!")

    except Exception as erro:
        conexao.rollback()
        print(f"Erro ao cadastrar produto: {erro}")

    finally:
        if cursor:
            cursor.close()            
```

```python
#Listar produto
@staticmethod
def listar_produtos(conexao):
    cursor = None
    try:
        cursor = conexao.cursor()

        sql = """SELECT id_produto, nome, tipo, preco, quantidade_disponivel
                 FROM produto
                 ORDER BY id_produto;"""

        cursor.execute(sql)
        resultados = cursor.fetchall()

        produtos = []

        for registro in resultados:
            produto = Produto(
                nome=registro[1],
                tipo=registro[2],
                preco=registro[3],
                quantidade_disponivel=registro[4]
            )
            produto.id_produto = registro[0]
            produtos.append(produto)

        return produtos

    except Exception as erro:
        print(f"Erro ao listar produtos: {erro}")
        return []

    finally:
        if cursor:
            cursor.close()
```
