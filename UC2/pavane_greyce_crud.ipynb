```python
class Cliente:
    def __init__(self, nome: str, telefone: str, email: str, tipo: str, id_cliente: int = None):
        self.id_cliente = id_cliente
        self.nome = nome
        self.telefone = telefone
        self.email = email
        self.tipo = tipo
    
    def __str__(self):
        return f"Cliente(nome={self.nome}, telefone={self.telefone}, email={self.email}, tipo={self.tipo})"
    
    def __repr__(self):
        return (f"Cliente(id_cliente={self.id_cliente!r}, " f"nome={self.nome!r}," f"telefone={self.telefone!r}," f"email={self.email!r}," f"tipo={self.tipo!r})"

     )
```

```python
#Cadastrar cliente
def inserir_cliente(self, conexao):
    cursor = None

    try:
        cursor = conexao.cursor()

        sql = """INSERT INTO cliente (nome, telefone, email, tipo)
                 VALUES (%s, %s, %s, %s)
                 RETURNING id_cliente;"""

        valores = (self.nome, self.telefone, self.email, self.tipo)

        cursor.execute(sql, valores)

        self.id_cliente = cursor.fetchone()[0]

        conexao.commit()

        print("Cliente cadastrado com sucesso!")

        return self.id_cliente

    except Exception as erro:
        conexao.rollback()

        print(f"Erro ao cadastrar cliente: {erro}")

        return None

    finally:
        if cursor is not None:
            cursor.close()
```

```python
#Listar clientes
@staticmethod
def listar_clientes(conexao):
    cursor = None

    try:
        cursor = conexao.cursor()

        sql = """SELECT id_cliente, nome, telefone, email, tipo
                 FROM cliente
                 ORDER BY id_cliente;"""

        cursor.execute(sql)
        resultados = cursor.fetchall()

        clientes = []

        for registro in resultados:
            cliente = Cliente(
                nome=registro[1],
                telefone=registro[2],
                email=registro[3],
                tipo=registro[4],
                id_cliente=registro[0]
            )

            clientes.append(cliente)

        return clientes

    except Exception as erro:
        print(f"Erro ao listar clientes: {erro}")
        return []

    finally:
        if cursor is not None:
            cursor.close()
```

```python
#Atualizar cliente
def atualizar_cliente(self, conexao):
    cursor = None

    try:
        cursor = conexao.cursor()

        sql = """UPDATE cliente
                  SET nome = %s, telefone = %s, email = %s, tipo = %s
                  WHERE id_cliente = %s;"""

        valores = (self.nome, self.telefone, self.email, self.tipo, self.id_cliente)

        cursor.execute(sql, valores)

        if cursor.rowcount == 0:
            conexao.rollback()
            print("Cliente não encontrado")
            return False
        
            conexao.commit()
            print("Cliente atualizado com sucesso!")
            return True

    except Exception as erro:
        conexao.rollback()

        print(f"Erro ao atualizar cliente: {erro}")

        return False

    finally:
        if cursor is not None:
            cursor.close()
```

```python
#Excluir cliente
@staticmethod
def excluir_cliente(id_cliente: int, conexao):
    cursor = None

    try:
        cursor = conexao.cursor()

        sql = """DELETE FROM cliente WHERE id_cliente = %s"""

        cursor.execute(sql, (id_cliente,))

        if cursor.rowcount == 0:
            conexao.rollback()
            print("Cliente não encontrado")
            return False

        conexao.commit()
        print("Cliente excluído com sucesso!")
        return True

    except Exception as erro:
        conexao.rollback()
        print(f"Erro ao excluir cliente: {erro}")
        return False

    finally:
        if cursor is not None:
            cursor.close()
```

```python
class Produto:
    id_produto: int
    nome: str
    tipo: str
    preco: float
    quantidade_disponivel: int

    def __init__(self, nome: str, tipo: str, preco: float, quantidade_disponivel: int):
        self.id_produto = None
        self.nome = nome
        self.tipo = tipo
        self.preco = preco
        self.quantidade_disponivel = quantidade_disponivel

    def __str__(self):
        return f"Produto(id_produto={self.id_produto}, nome={self.nome}, tipo={self.tipo}, preco={self.preco}, quantidade_disponivel={self.quantidade_disponivel})"

    def __repr__(self):
        return self.__str__()
```

```python
#inserir produto
def inserir_produto(self,conexao):
    cursor = None
    try:
        cursor = conexao.cursor()

        sql = """INSERT INTO produto (nome, tipo, preco, quantidade_disponivel)
                 VALUES (%s, %s, %s, %s)
                 RETURNING id_produto;"""

        valores = (self.nome, self.tipo, self.preco, self.quantidade_disponivel)

        cursor.execute(sql, valores)

        self.id_produto = cursor.fetchone()[0]

        conexao.commit()

        print("Produto cadastrado com sucesso!")

    except Exception as erro:
        conexao.rollback()
        print(f"Erro ao cadastrar produto: {erro}")

    finally:
        if cursor:
            cursor.close()            
```

```python
from datetime import date

class Encomenda:
    id_encomenda: int 
    id_cliente: int
    data_pedido: date
    data_pagamento: date
    data_retirada: date
    descricao: str
    
    def __init__(self, id_cliente: int, data_pedido: date, data_pagamento: date, data_retirada: date, descricao: str):
        self.id_encomenda = None
        self.id_cliente = id_cliente
        self.data_pedido = data_pedido
        self.data_pagamento = data_pagamento
        self.data_retirada = data_retirada
        self.descricao = descricao
    
    def __str__(self):
        return f"Encomenda(id_encomenda={self.id_encomenda}, id_cliente={self.id_cliente}, data_pedido={self.data_pedido}, data_pagamento={self.data_pagamento}, data_retirada={self.data_retirada}, descricao={self.descricao})"

    def __repr__(self):
        return self.__str__()
```

```python
#Inserir_encomenda
def inserir_encomenda(self,conexao):
    cursor = None
    try:
        cursor = conexao.cursor()

        sql = """INSERT INTO encomenda (id_cliente, data_pedido, data_pagamento, data_retirada, descricao)
                 VALUES (%s, %s, %s, %s, %s)
                 RETURNING id_encomenda;"""

        cursor.execute(sql, (self.id_cliente, self.data_pedido, self.data_pagamento, self.data_retirada, self.descricao))

        self.id_encomenda = cursor.fetchone()[0]
        
        conexao.commit()
        return True
     
    except Exception as erro:
        conexao.rollback()
        print(f"Erro ao cadastrar encomenda: {erro}")
        return False

    finally:
        if cursor:
            cursor.close()
```

```python
# Consultar encomenda
@staticmethod
def consultar_encomenda(conexao, id_encomenda):
    cursor = None

    try:
        cursor = conexao.cursor()

        sql = """
            SELECT id_encomenda, id_cliente, data_pedido,
                   data_pagamento, data_retirada, descricao
            FROM encomenda
            WHERE id_encomenda = %s;
        """

        cursor.execute(sql, (id_encomenda,))
        registro = cursor.fetchone()

        if registro:
            encomenda = Encomenda(
                id_cliente=registro[1],
                data_pedido=registro[2],
                data_pagamento=registro[3],
                data_retirada=registro[4],
                descricao=registro[5]
            )

            encomenda.id_encomenda = registro[0]
            return encomenda

        return None

    except Exception as erro:
        print(f"Erro ao consultar encomenda: {erro}")
        return None

    finally:
        if cursor:
            cursor.close()
```

```python
class Item_Encomenda():
    id_produto: int 
    id_encomenda: int
    quantidade: int
    preco: float

    def __init__(self, id_produto:int, id_encomenda:int, quantidade:int, preco:float):
        self.id_produto=id_produto
        self.id_encomenda=id_encomenda
        self.quantidade=quantidade
        self.preco=preco

    def __str__(self):
        return f"Item_Encomenda(id_produto={self.id_produto}, id_encomenda={self.id_encomenda}, quantidade={self.quantidade}, preco={self.preco})"
    
    def __repr__(self):
        return self.__str__()
```

```python
# Inserir item_encomenda
def inserir_item_encomenda(self, conexao):
    cursor = None

    try:
        cursor = conexao.cursor()
        sql = """INSERT INTO item_encomenda (id_produto, id_encomenda, quantidade, preco) VALUES (%s, %s, %s, %s);"""

        cursor.execute(sql, (
            self.id_produto,
            self.id_encomenda,
            self.quantidade,
            self.preco
        ))

        conexao.commit()
        return True

    except Exception as erro:
        conexao.rollback()
        print(f"Erro ao inserir item da encomenda: {erro}")
        return False

    finally:
        if cursor:
            cursor.close()
```

```python
# Consultar item_encomenda
@staticmethod
def consultar_item_encomenda(conexao, id_produto, id_encomenda):
    cursor = None

    try:
        cursor = conexao.cursor()

        sql = """
            SELECT id_produto, id_encomenda, quantidade, preco
            FROM item_encomenda
            WHERE id_produto = %s AND id_encomenda = %s;
        """

        cursor.execute(sql, (id_produto, id_encomenda))
        registro = cursor.fetchone()

        if registro:
            return Item_Encomenda(
                registro[0],
                registro[1],
                registro[2],
                registro[3]
            )

        return None

    except Exception as erro:
        print(f"Erro ao consultar item da encomenda: {erro}")
        return None

    finally:
        if cursor:
            cursor.close()
```

```python
# Listar itens_encomenda
@staticmethod
def listar_itens_encomenda(conexao):
    cursor = None

    try:
        cursor = conexao.cursor()

        sql = """
            SELECT id_produto, id_encomenda, quantidade, preco
            FROM item_encomenda
            ORDER BY id_encomenda;
        """

        cursor.execute(sql)
        resultados = cursor.fetchall()

        itens = []

        for registro in resultados:
            item = Item_Encomenda(
                registro[0],
                registro[1],
                registro[2],
                registro[3]
            )

            itens.append(item)

        return itens

    except Exception as erro:
        print(f"Erro ao listar itens da encomenda: {erro}")
        return []

    finally:
        if cursor:
            cursor.close()
```

```python
# Atualizar item_encomenda
def atualizar_item_encomenda(self, conexao):
    cursor = None

    try:
        cursor = conexao.cursor()

        sql = """
            UPDATE item_encomenda
            SET quantidade = %s,
                preco = %s
            WHERE id_produto = %s
            AND id_encomenda = %s;
        """

        cursor.execute(sql, (
            self.quantidade,
            self.preco,
            self.id_produto,
            self.id_encomenda
        ))

        conexao.commit()
        return cursor.rowcount > 0

    except Exception as erro:
        conexao.rollback()
        print(f"Erro ao atualizar item da encomenda: {erro}")
        return False

    finally:
        if cursor:
            cursor.close()
```

```python
# Excluir item_encomenda
@staticmethod
def excluir_item_encomenda(conexao, id_produto, id_encomenda):
    cursor = None

    try:
        cursor = conexao.cursor()

        sql = """
            DELETE FROM item_encomenda
            WHERE id_produto = %s
            AND id_encomenda = %s;
        """

        cursor.execute(sql, (id_produto, id_encomenda))

        conexao.commit()
        return cursor.rowcount > 0

    except Exception as erro:
        conexao.rollback()
        print(f"Erro ao excluir item da encomenda: {erro}")
        return False

    finally:
        if cursor:
            cursor.close()
```
